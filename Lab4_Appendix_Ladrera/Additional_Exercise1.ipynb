```python
last_name = "Ladrera"
seed_num = 2

sources = [f"{last_name}_T1", f"{last_name}_T2", f"{last_name}_T3", f"{last_name}_T4", f"{last_name}_T5"]
records = [
    {"source": sources[0], "category": "TEMPERATURE", "value": seed_num * 5},
    {"source": sources[1], "category": "VOLTAGE", "value": seed_num * 12},
    {"source": sources[2], "category": "CURRENT", "value": seed_num * 3},
    {"source": sources[0], "category": "TEMPERATURE", "value": seed_num * 5} # Duplicate
]

print("1. Generated:", records)

organized = {r["source"]: r["value"] for r in records}
print("2. Organized:", organized)


unique_data = set(tuple(r.items()) for r in records)
print("3. Unique Count:", len(unique_data))


total = sum(r["value"] for r in records)
print("4. Total Value Sum:", total)

ordered = sorted(records, key=lambda x: x["value"])
print("5 & 6. Sorted Results:", ordered)

print("7. Final Output: Successfully organized and summarized test dataset.")
```
