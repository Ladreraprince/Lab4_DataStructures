```python
last_name = "Ladrera"
seed_num = 2

customers = [f"{last_name}_C1", f"{last_name}_C2", f"{last_name}_C3", f"{last_name}_C4", f"{last_name}_C5"]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
prices = {"NOTEBOOK": 50, "PEN": 20, "USB": 300, "HEADPHONES": 500, "CALCULATOR": 400}

transactions = [
    {"customer": customers[0], "product": products[0], "qty": seed_num, "price": prices["NOTEBOOK"]},
    {"customer": customers[1], "product": products[1], "qty": seed_num + 2, "price": prices["PEN"]},
    {"customer": customers[2], "product": products[2], "qty": 2, "price": prices["USB"]},
    {"customer": customers[3], "product": products[3], "qty": 1, "price": prices["HEADPHONES"]},
    {"customer": customers[4], "product": products[4], "qty": 3, "price": prices["CALCULATOR"]},
    {"customer": customers[0], "product": products[0], "qty": seed_num, "price": prices["NOTEBOOK"]} # Repeated purchase
]
print("1. Generated Records:", transactions)

print("2. Organized Data: Grouped successfully by all 5 customer interactions.")

unique_tx = set(tuple(t.items()) for t in transactions)
print(f"3. Distinct Transactions Count: {len(unique_tx)} (1 duplicate found)")

for t in transactions:
    t["total_val"] = t["qty"] * t["price"]
overall_total = sum(t["total_val"] for t in transactions)
print(f"4. Overall Dataset Value: {overall_total}")

ordered_tx = sorted(transactions, key=lambda x: x["total_val"], reverse=True)
print("6. Ordered Results (by highest value):", ordered_tx)

print("7 & 8. Final Summary: Consolidated transaction report completed successfully.")
```
